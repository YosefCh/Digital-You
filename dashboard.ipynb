### How to compute sleep in SQL (later)
- **Time in bed (minutes)**:
  - `EXTRACT(EPOCH FROM (wake_time - bedtime))/60`
- **Estimated total minutes asleep**:
  - `time_in_bed_minutes - COALESCE(awake_gap_minutes, 0)`
- **Uninterrupted flag (derived)**:
  - `awake_gap_minutes IS NULL OR awake_gap_minutes = 0`

### Late night followed by an early night (edge case)
If `bedtime` is after midnight on day X, it will be recorded as the actual date, despite being inconsistent with the user's perception of "date". For example, if a user goes to bed after Sunday Jan 1st at 1am on Jan 2nd, it will be recorded as `bedtime = '2024-01-02 01:00:00'`. If on Monday they go to bed at 11pm, it will be recorded as `bedtime = '2024-01-02 23:00:00'`. This can lead to confusion when analyzing sleep patterns by date, since the Jan 2nd sleep will be associated with Jan 2nd instead of Jan 1st. 

This must be addressed either in the insert or in the select when analyzing sleep. We also must take into cosideration that we still want a constraint to catch duplicate entries. One solution is to have a separate `sleep_date` column that represents the user's perception of the date for that sleep session, and enforce uniqueness on `(user_id, sleep_date)`. The `bedtime` and `wake_time` can still be stored as timestamps without modification.
